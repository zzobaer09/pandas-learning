## Accessing Data with Pandas

### df.loc[] df.iloc[]

```python
coffee_df
```

```python
coffee_df.loc[0]
```

```python
coffee_df.loc[[2,3,6]]
```

```python
coffee_df.loc[5:9,["Coffee Type","Units Sold"]]
```

```python
coffee_df.iloc[5:9,[2,1]]
```

Pasted image.png

Memory Copy vs. View: Standard row/column slices (like coffee_df.iloc[5:9, 1:3]) return an internal memory view when possible, whereas passing an explicit index list ([2, 1]) forces pandas to allocate a brand-new memory copy.

```python
coffee_df.loc[coffee_df['Day'] == 'Monday', 'Units Sold']
```

### df.index

```python
coffee_df1 = coffee_df.copy()
coffee_df1.index = coffee_df1.Day
coffee_df1.loc["Monday":"Thursday","Coffee Type":"Units Sold"]
```

### Setting DataFrame Values

```python
coffee_df1.iloc[1:4,2] = 100
coffee_df1
```

## Filtering Data

```python
orexdf.head()
```

```python
orexdf.tail()
```

```python
orexdf.info()
```

```python
orexdf_height_mask = orexdf["height_cm"]>215
orexdf.loc[orexdf_height_mask,["name","height_cm","born_region"]]
```

```python
orexdf_country_mask = orexdf.loc[:,"born_country"]=="RUS"
orexdf[(orexdf_height_mask & orexdf_country_mask)]
```

Idiom & Speed: orexdf["col"] is standard and preferred for single-column access. It is slightly faster than .loc[:, "col"] because it bypasses the 2D row/column parsing logic of the .loc indexer.
