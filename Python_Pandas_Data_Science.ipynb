### custom function

```python
orexdf1["height_category"] = orexdf1["height_cm"].apply(lambda x: "NOT ENOUGH DATA" if pd.isna(x) else("Short" if x<165 else ("Average" if x<185 else "Tall")))
orexdf1[["name","height_cm","height_category"]]
```

```python
def categorize_athletes(row):
    if pd.isna(row["height_cm"]) or pd.isna(row["weight_kg"]):
        return "Not Enough Data"
    if row["height_cm"]<175 and row["weight_kg"]<70:
        return "Lightweight"
    elif row["height_cm"]<185 and row["weight_kg"]<80:
        return "Middleweight"
    else: return "Heavyweight"
```

```python
orexdf1["category"] = orexdf1.apply(categorize_athletes,axis=1)
```

```python
orexdf1[["first_name","last_name","height_category","height_cm","weight_kg","category","born_year"]]
```
